```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langgraph.graph import StateGraph,START,END
from pydantic import BaseModel
```

```python
load_dotenv()
```

```python
llm = ChatGroq(model='openai/gpt-oss-120b')
```

```python
class report(BaseModel):
    statement: str
    outline: str = ''
    report: str = ''
```

```python
def create_outline(state: report)-> dict:
    prompt = f'generate the outline for this topic {state.statement}'

    outline = llm.invoke(prompt).content
    state.outline = outline

    return state
```

```python
def create_report(state: report)-> dict:
       

    report_prompt = f'create a detailed blog on this topic {state.statement} using this outline {state.outline}'
    report = llm.invoke(report_prompt).content

    state.report = report
    return state
```

```python
graph = StateGraph(report)

graph.add_node('create_outline',create_outline)
graph.add_node('create_report',create_report)

graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_report')
graph.add_edge('create_report',END)

workflow = graph.compile()
```

```python
initial_state = {'statement':'evolution of AI'}

final_state = workflow.invoke(initial_state)
```

```python
print(final_state['statement'])
print(final_state['outline'])
print(final_state['report'])
```
