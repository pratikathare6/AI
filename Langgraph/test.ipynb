```python
from langgraph.graph import START,END,StateGraph
from typing import TypedDict
```

```python
class char_convert(TypedDict):
    word : str
```

```python
def string_change(state : char_convert)->char_convert:

    state['word'] = state['word'].upper()
     
    return state

```

```python
graph = StateGraph(char_convert)
graph.add_node('string_change',string_change)
graph.add_edge(START,'string_change')
graph.add_edge('string_change',END)

workflow = graph.compile()
```

```python
initial_state = {'word':'this is text'}

final_state = workflow.invoke(initial_state)

final_state
```
