```python
import os 
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from dotenv import load_dotenv
```

```python
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["LANGSMITH_API_KEY"] = os.getenv("LANGSMITH_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"] = "true" #very important for enable tracing this is actual langsmith
```

```python
prompt = ChatPromptTemplate.from_messages(
    [
        ("system","you are a helpfull assistant. please response to the used bsed on the provided context only"),
        ("user","Question:{question}\nContext:{context}")
    ]
)
```

```python
llm = ChatGroq(model='openai/gpt-oss-20b')
output_parser = StrOutputParser()
```

```python
chain = prompt | llm | output_parser
```

```python
question = "can you summarise this story?"

context = """The Thirsty CrowThe day was very hot. The crow felt weak and tired. He could not find water anywhere.Suddenly, he saw a small pitcher on the ground. He flew down quickly to look inside. There was a little water at the bottom. 
    But the crow's beak could not reach it.He tried to push his head inside. He failed. He tried to break the pitcher. 
    It was too hard.Then, the clever crow saw small pebbles on the ground. He had a good idea.
    He picked up a pebble with his beak. He dropped it into the pitcher. Plop.He picked up another pebble and dropped it in. Plop.He worked hard and put many pebbles inside.
      Slowly, the water rose to the top.The crow bent his head and drank the cool water. 
      He felt happy and flew away into the bright sky."""

```

```python
response = chain.invoke({"question":question, "context":context})
print(response)
```
