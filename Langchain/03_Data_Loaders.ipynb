```python
import os
from dotenv import load_dotenv,find_dotenv
_ = load_dotenv(find_dotenv())
groq_api_key = os.environ['GROQ_API_KEY']
```

```python
from langchain_groq import ChatGroq

chat_model = ChatGroq(model='llama-3.3-70b-versatile')
```

```python
messages = [
    ("system","you are helpfull asistant"),
    ("human","tell me about jj")
]

res = chat_model.invoke(messages)
print(res.content)
```

## above person jj information is not available in the model so we can provide the information using **Data Loader**

```python
#txt data loading
from langchain_community.document_loaders import TextLoader

loader = TextLoader("Data/person.txt") 
loaded_data = loader.load()
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages(

    [
        ("human","Answer this question {question}, here is some extra {context}")
    ]
)

messages = chat_template.format_messages(

    name = "jj",
    question = "tell me about this person",
    context = loaded_data
)

res = chat_model.invoke(messages)
print(res.content)
```
